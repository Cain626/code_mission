# 探索性分析（EDA）

先运行第 1 格定位项目根目录，再运行第 2 格加载数据。
把 CSV 放在 `data/raw/` 下即可自动识别（多个文件时取第一个，或改 `csv` 变量指定）。

```python
import sys
from pathlib import Path

# 定位仓库根目录（含 src/ 的那一层），保证 src 可导入
ROOT = Path.cwd()
while not (ROOT / "src").exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

import pandas as pd
from src.io_utils import load_csv
from src import plotting as P

pd.set_option("display.width", 160)
pd.set_option("display.max_columns", 50)
print("项目根目录:", ROOT)
print("中文字体:", P.setup_style())
```

```python
candidates = sorted((ROOT / "data" / "raw").glob("*.csv"))
csv = candidates[0] if candidates else None          # 想指定文件就写死路径

df = load_csv(csv) if csv else None
df.head() if df is not None else print("data/raw/ 下还没有 CSV")
```

```python
if df is not None:
    print(df.shape)
    display(df.dtypes.to_frame("类型"))
    display(df.isna().mean().mul(100).round(1).to_frame("缺失率%"))
```

```python
# 示例：挑一个类别列画分布
if df is not None:
    col = df.columns[2]
    fig, ax = P.bar_top(df[col], top=10)
    P.save_fig(fig, f"eda_{col}")
```
